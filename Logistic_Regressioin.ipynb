```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/Social_Network_Ads.csv')
```

```python
df.drop(['Gender','User ID','EstimatedSalary'],axis=1,inplace=True)
```

```python
df.sample(2)
```

```python
plt.figure(figsize=(4,3))
sns.scatterplot(x='Age',y='Purchased',data=df,hue='Purchased')
plt.show()
```

```python
plt.figure(figsize=(4,3))
sns.scatterplot(x='Age',y='Purchased',data=df)
plt.show()
```

```python
x=df[['Age']]
y=df['Purchased']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
model.predict([[79]])
```

```python
plt.figure(figsize=(4,3))
sns.scatterplot(x='Age',y='Purchased',data=df)
sns.lineplot(x='Age',y=model.predict(x),data=df,color='red')
plt.show()
```

```python
plt.figure(figsize=(4,3))
sns.scatterplot(x='Age',y='Purchased',data=df)
sns.scatterplot(x='Age',y=model.predict(x),data=df,color='red')
plt.show()
```

```python
y_probability = model.predict_proba(x_test)

print("\nProbabilities:")
print(y_probability)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report,mean_absolute_error

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
print("\nMean Absolute Error:", mean_absolute_error(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
accuracy = accuracy_score(y_test, y_pred)*100

print("\nAccuracy:", accuracy)
```
