```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/placement.csv')
```

```python
df
```

```python
df.isnull().sum()
```

```python
x=df.iloc[:,:-1]
y=df['package']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
```

```python
lr=LinearRegression()
dtr=DecisionTreeRegressor()
svr=SVR()
```

```python
lr.fit(x_train,y_train)
lr.score(x_train,y_train)*100,lr.score(x_test,y_test)*100
```

```python
dtr.fit(x_train,y_train)
dtr.score(x_train,y_train)*100,dtr.score(x_test,y_test)*100
```

```python
svr.fit(x_train,y_train)
svr.score(x_train,y_train)*100,svr.score(x_test,y_test)*100
```

```python
from sklearn.ensemble import VotingRegressor
```

```python
voting_regressor=VotingRegressor([('lr',lr),('dtr',dtr),('svr',svr)])
voting_regressor.fit(x_train,y_train)
voting_regressor.score(x_train,y_train)*100,voting_regressor.score(x_test,y_test)*100
```

```python
df=pd.DataFrame({"Linear Regression":lr.predict(x_test),"Decision Tree Regression":dtr.predict(x_test),"Support Vector Regression":svr.predict(x_test),"Voting Regressor":voting_regressor.predict(x_test)})
```
