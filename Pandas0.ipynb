```python
import numpy as np
import pandas as pd
data= pd.read_csv("/content/customer_purchase_data.csv")
```

```python
data.head(3)
```

```python
data.groupby(["Gender","Country"]).count()
```

```python
data.isnull().sum()
```

```python
data.dtypes
```

```python
data["CustomerName"].dtype
data["CustomerName"]=data["CustomerName"].fillna(data["CustomerName"].mode()[0])
data["Email"]=data["Email"].fillna(data["Email"].mode()[0])
data["ReviewScore"]=data["ReviewScore"].fillna(data["ReviewScore"].mode()[0])
mean_age=data["Age"].mean()
mean_purchase_amount=data["PurchaseAmount"].mean()
data["Age"]=data["Age"].fillna(mean_age)
data["PurchaseAmount"]=data["PurchaseAmount"].fillna(mean_purchase_amount)
```

```python
data.describe()
```

```python
data.isnull().sum()
```

```python
data["PurchaseAmount"].value_counts()
```

```python
data.groupby("PaymentMethod").agg({"PurchaseAmount":["mean","sum","max","min"],
                                  "ProductCategory":"count"  })
```

```python
bins=[0,18,25,35,55,70]
labels=["Teens","Young","Adults","Mid","Aged"]
data["AgeGroup"]=pd.cut(data["Age"],bins=bins,labels=labels)
data
```

```python
data["AgeGroupQcut"]=pd.qcut(data["Age"],5,labels=labels)
data
```

```python
scores = np.random.randint(0, 101, 20)
df = pd.DataFrame({"Score": scores})
bins = [0, 59, 75, 89, 100]
labels = ["F", "C", "B", "A"]
df["Grade"] = pd.cut(df["Score"], bins=bins, labels=labels)
print("Scores with Grades:")
print(df)
print("\nGrade Counts:")
print(df["Grade"].value_counts())
```
