```python
df=pd.read_csv('tip.csv')
df
```

```python
plt.scatter(
    x = df['total_bill'],y = df['tip'],
    s=df['total_bill'] * 3,
    marker='o',
    facecolor='red',
    edgecolor='black'
)

plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

```python
# Histogram

df=pd.read_csv('pakistani_cricketers_scores.csv')
```

```python
plt.hist([df['Babar Azam'],df['Shaheen Afridi'],df['Mohammad Rizwan']],edgecolor="black",label=['Baber','Shaheen','Rizwan'],histtype='bar')
plt.legend()
plt.xlabel('Score')
plt.show()
```

```python
plt.hist([df['Babar Azam'],df['Mohammad Rizwan']],label=['Baber','Rizwan'],bins=10, alpha=0.5)
plt.legend()
plt.xlabel("Matches")
plt.ylabel("Score")
plt.show()
```

```python
plt.hist(df['Shaheen Afridi'],bins=5,edgecolor='black')
plt.xlabel('Runs')
plt.ylabel('Frequency')
plt.title('Shaheen Afridi')
plt.show
```

```python
df=pd.read_csv('pakistani_cricketers_yearly_avg2.csv')
```

```python
df
```

```python
plt.bar(df['Batsman'],df['2021'],label='2021',width=0.6)
plt.show()
```

```python
plt.bar(df['Batsman'],df['2021'],label='2021')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.bar(df['Batsman'],df['2021'],label='2021')
plt.bar(df['Batsman'],df['2023'],label='2023',bottom=df['2021'])
plt.bar(df['Batsman'],df['2025'],label='2025',bottom=df['2021']+df['2023'])
plt.legend()
plt.xticks(rotation=45)
plt.xlabel('Batsman')
plt.ylabel('Runs')
plt.show()
```

```python
plt.barh(df['Batsman'],df['2026'],label='2026')
plt.legend()
plt.show()
```

```python
plt.bar(df['Batsman'],df['2021'],label='2021')
plt.bar(df['Batsman'],df['2022'],label='2022',bottom=df['2021'])
plt.bar(df['Batsman'],df['2023'],label='2023',bottom=df['2021']+df['2022'])
plt.bar(df['Batsman'],df['2024'],label='2024',bottom=df['2021']+df['2022']+df['2023'])
plt.bar(df['Batsman'],df['2025'],label='2025',bottom=df['2021']+df['2022']+df['2023']+df['2024'])
plt.bar(df['Batsman'],df['2026'],label='2026',bottom=df['2021']+df['2022']+df['2023']+df['2024']+df['2025'])
plt.legend()
plt.xticks(rotation=60)
plt.xlabel('Batsman')
plt.ylabel('Runs')
plt.title('Average Runs for year 2021,2022,2023,2024,2025,2026')
plt.show()
```
