Error Handling

```python
try:
  Name=input("Enter your name ")
  Semester=int(input("Enter your semester "))
  total_sub=int(input(f"Enter total number of subject in semester {Semester}     "))
  total_marks=0
  for i in range(total_sub):
    marks=int(input(f"Enter marks for subject  {i+1}   "))
    if marks<0 or marks>100:
      raise ValueError("Marks must be between 0 and 100")
    total_marks+=marks
  percentage=total_marks/(total_sub*100)*100

except ValueError as ae:
  print("Value Error ")
except Exception as e:
  print("Error occured")

else:
  print("\n")
  print(f"Name:  {Name}")
  print(f"Semester:  {Semester}")
  print(f"Total Marks:  {total_marks} out of {total_sub*100}")
  print(f"Percentage:  {percentage}%98")
  if percentage>85:
    print("Grade A:")
  elif percentage>=75:
    print("Grade B:")
  elif percentage>=65:
    print("Grade C:")
  elif percentage>=50:
    print("Grade D:")
  else:
    print("Fail")
finally:
  print("Program executed sucessfully7")
```

Reshape

```python
arr5= np.arange(12)
arr6=arr5.reshape(2,6)
print(arr6)

arr7=arr6.flatten() #Converts array to 1-D, Creates a NEW copy of the data and If you change array, original array does NOT change.
arr7[3]=70
print(f"Flatten using flatten {arr7}")

arr8=arr6.ravel() #Converts array to 1-D ,Does NOT create a new copy (view of original array) and If you change r, original array MAY change.
arr8[2]=49
print(f"Flatten using Ravel {arr8}")
```
