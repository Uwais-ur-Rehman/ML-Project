```python
dep=pd.read_csv('/content/Book1.csv' ,index_col='Vacancies')
dep1=dep.value_counts()
print(f"Count Values: {dep1}")
print(f"\nSort Values: {dep1.sort_values(ascending=False)}")
print(f"\nHighest Values with Deptt: {dep1.sort_values(ascending=False).head(1)}")
print(f"\nHighest Values only: {dep1.sort_values(ascending=False).head(1).values}")
print(f"\npermanent save Values: {dep1.sort_values(inplace=True)}")
print(dep1)

print('\n')
print(f"\nSort index: {dep1.sort_index()}")
print("\n")
print(f"sum is: {dep1.sum()}")
print(f"product is: {dep1.product()}")
print(f"mean is: {dep1.mean()}")
print(f"median  is: {dep1.median()}")
print(f"mode is: {dep1.mode()}")
print(f"Standard Deviation is: {dep1.std()}")
print(f"Variance is: {dep1.var()}")
print(f"max is: {dep1.max()}")
print(f"min is: {dep1.min()}")
print(f"Describe is: {dep1.describe()}")
dep1[1]=600
print(f"After alteration{dep1}")
dep1['sst']=90
print(dep1)
```
