```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/Placement.csv')
df.sample(2)
```

```python
x=df.drop(columns=['Placement','Student_ID'])
y=df['Placement']
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x=pd.DataFrame(sc.fit_transform(x),columns=x.columns)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
from sklearn.tree import DecisionTreeClassifier
dt=DecisionTreeClassifier()
dt.fit(x_train,y_train)
```

```python
dt.score(x_test,y_test)*100
```

```python
dt.score(x_train,y_train)*100
```

```python
y_pred=dt.predict(x_test)
```

```python
dt.predict([[1.0,4]])
```

```python
from sklearn.tree import plot_tree
plt.figure(figsize=(10,10))
plot_tree(dt)
plt.show()
```

```python
dt1=DecisionTreeClassifier(max_depth=2)
dt1.fit(x_train,y_train)
```

```python
dt1.score(x_train,y_train)*100
```

```python
dt1.score(x_test,y_test)*100
```

```python
sns.scatterplot(x='CGPA',y='IQ',data=df,hue='Placement')
plt.show()
```

```python
from mlxtend.plotting import plot_decision_regions
plt.figure(figsize=(5,5))
plot_decision_regions(x.values,y.to_numpy(),dt1)
plt.show()
```

```python
for i in range (1,20):
  dt2=DecisionTreeClassifier(max_depth=i)
  dt2.fit(x_train,y_train)
  print(i,dt2.score(x_train,y_train)*100,dt2.score(x_test,y_test)*100,dt2.score(x_train,y_train)*100-dt2.score(x_test,y_test)*100)
```
