```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/dataset_study.csv')
df
```

```python
df.isnull().sum()
```

```python
x=df[['study_hours']]
y=df['grade']
```

```python
y.sample(3)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
from sklearn.linear_model import LinearRegression
model= LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred= model.predict(x_test)
```

```python
df1 = pd.DataFrame({
    "Actual Marks": y_test.values,
    "Predicted Marks": y_pred
})

print(df1)
```

```python
print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
hours = [[6]]
prediction = model.predict(hours)

print("Predicted marks for 6 hours:", prediction)
```

```python
import pickle

with open("model.pkl", "wb") as file:
    pickle.dump(model, file)
```

```python
sns.scatterplot(x="study_hours",y="grade",data=df)
plt.plot(x_test,y_pred,color='red')
plt.xlabel("Study Hours")
plt.ylabel("Grade")
plt.title("Study Hours vs Grade")
plt.legend(["origional data","Predicted data"])
plt.show()
plt.savefig("study_hours_vs_grade.png")
```
