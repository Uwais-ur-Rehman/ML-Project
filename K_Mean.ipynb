```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/iris_unsupervised.csv')
df.head(2)
```

```python
sns.pairplot(df)
plt.show()
```

```python
from sklearn.cluster import KMeans
```

```python
wcss=[]
for i in range(2,21):
  kn=KMeans(n_clusters= i, init="k-means++")
  kn.fit(df)
  wcss.append(kn.inertia_)
```

```python
plt.figure(figsize=(12,6))
plt.plot([i for i in range(2,21)],wcss,marker='o')
plt.xlabel("No. of cluster")
plt.ylabel("WCSS")
plt.xticks([i for i in range(2,21)])
plt.grid()
plt.show()
```

```python
knn=KMeans(n_clusters=3)
df['Predict']=knn.fit_predict(df)
```
