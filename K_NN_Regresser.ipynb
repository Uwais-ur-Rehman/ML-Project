```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/Salary Data.csv')
df.sample(2)
```

```python
df.isnull().sum()
```

```python
print(df[df.isnull().any(axis=1)])
```

```python
df.dropna(inplace=True)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df['Education Level'].value_counts()
```

```python
df['Gender'].value_counts()
```

```python
x=df.drop(columns=['Job Title','Salary'])
y=df['Salary']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
from sklearn.compose import ColumnTransformer
numerical=['Age','Years of Experience']
ordinal=['Education Level']
nominal=['Gender']
preprocessor=ColumnTransformer([
    ('num',StandardScaler(),numerical),
    ('ord',OrdinalEncoder(),ordinal),
    ('nom',OneHotEncoder(sparse_output=False,drop='first',handle_unknown="ignore"),nominal)
])
x_train=preprocessor.fit_transform(x_train)
x_test=preprocessor.transform(x_test)
```

```python
from sklearn.neighbors import KNeighborsRegressor
knnr=KNeighborsRegressor(n_neighbors=18)
knnr.fit(x_train,y_train)
```

```python
knnr.score(x_test,y_test)*100
```

```python
knnr.score(x_train,y_train)*100
```

```python
for i in range(1,30):
  knnr=KNeighborsRegressor(n_neighbors=i)
  knnr.fit(x_train,y_train)
  print(i,knnr.score(x_train,y_train)*100,knnr.score(x_test,y_test)*100,knnr.score(x_train,y_train)*100-knnr.score(x_test,y_test)*100)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
y_pred=knnr.predict(x_test)
```

```python
print("Mean Absolute Error:", mean_absolute_error(y_test, y_pred))
print("Mean Squared Error:", mean_squared_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred)*100)
```
