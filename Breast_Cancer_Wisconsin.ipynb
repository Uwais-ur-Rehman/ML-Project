```python
 import numpy as np
 import pandas as pd
 import matplotlib.pyplot as plt
 import seaborn as sns
```

```python
df=pd.read_csv('/content/Breast Cancer Wisconsin.csv')
```

```python
df.shape
```

```python
df.drop(columns=['Unnamed: 32','id'],axis=1,inplace=True)
```

```python
df.shape
```

```python
df.sample(2)
```

```python
x=df.drop(columns=['diagnosis'])
y=df['diagnosis']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
from sklearn.naive_bayes import GaussianNB
model=GaussianNB()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
model.score(x_test,y_test)*100
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,precision_score,recall_score,f1_score
```

```python
cm=confusion_matrix(y_test,y_pred)
cm
```

```python
sns.heatmap(cm,annot=True)
plt.show()
```

```python
print('Accuracy Score:',accuracy_score(y_test,y_pred)*100)
print('Precision Score:',precision_score(y_test,y_pred,pos_label='M')*100)
print('Recall Score:',recall_score(y_test,y_pred,pos_label='M')*100)
print('F1 Score:',f1_score(y_test,y_pred,pos_label='M')*100)
```
