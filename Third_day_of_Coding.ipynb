Creating DataFrame using list

```python
import numpy as np
import pandas as pd

student_marks=[
    [77,88,77,66,77,87],
    [66,77,55,44,88,98],
    [98,87,89,90,87,66],
    [67,73,91,81,72,59]
]
marks=pd.DataFrame(student_marks,columns=['c++','python','java','c','php','Sql'])
print(marks)
```

```python
df_student_marks = pd.DataFrame(student_marks, columns=['c++','python','java','c','php','Sql'])
df_student_marks = df_student_marks.rename(columns={'c++':'C++','python':'PYTHON','java':'JAVA','c':'C','php':'PHP','Sql':'SQL'})
print(df_student_marks)
```

Creating DataFrame using Dict

```python
student_marks1={
    'Name':['Ali','Ahmad','Ayesha','Adil','Adnan','Abeer'],
    'Java':[77,88,87,66,77,87],
    'C++':[67,73,91,81,72,59],
    'C':[66,77,86,94,88,98],
    'Python':[98,87,89,90,87,66],
    'Obtained':[0,0,0,0,0,0],
    'Total':[0,0,0,0,0,0],
    'Percentage':[0,0,0,0,0,0]
}
marks1=pd.DataFrame(student_marks1)
marks1.set_index('Name',inplace=True)
marks1['Obtained']=marks1[['Java','C++','C','Python']].sum(axis=1)
marks1['Total']=marks1[['Java','C++','C','Python']].count(axis=1)*100
marks1['Percentage']=(marks1['Obtained']/marks1['Total'])*100
print(marks1)
print(f"\n Marks:\n {marks1.loc[[marks1['Percentage'].idxmax()]]}\n")
print(marks1['Percentage'])
```

```python
def grade_cal(value):
  if value>=85:
    return "A"
  elif value>=75:
    return "B"
  elif value>=65:
    return "C"
  elif value>=55:
    return "D"
  elif value>=50:
    return "Pass"
  else:
    return "Fail"
```

```python
marks1['Grade']= marks1['Percentage'].apply(grade_cal)
print(marks1)
```

```python
marks1.value_counts()
```

```python
marks1.rename(columns={'C':'JavaScript'},inplace=True)
print(marks1)
```

```python
#marks1['Not a Number'].hasnans
#print(marks1['Not a Number'][marks1['Not a Number'].isnull()])
marks1.dropna(how='any')
print(marks1)
```
