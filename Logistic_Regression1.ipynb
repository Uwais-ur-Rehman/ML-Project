```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/Placement.csv')
```

```python
df
```

```python
df.isnull().sum()
```

```python
sns.scatterplot(x='CGPA',y='IQ',data=df,hue='Placement')
sns.lineplot(x='CGPA',y='IQ',data=df,color='red',hue='Placement')
plt.show()
```

```python
df.drop(['Student_ID'],axis=1,inplace=True)
```

```python
x=df[['CGPA','IQ']]
y=df['Placement']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=40)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

```python
y_pred= model.predict(x_test)
```

```python
model.score(x_test,y_test)*100
```

```python
model.predict([[9,102]])
```

```python
plt.figure(figsize=(4,3))
sns.scatterplot(x='CGPA',y='Placement',data=df)
sns.lineplot(x='CGPA',y=model.predict(x),data=df,color='red')
plt.show()
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
plot_decision_regions(x.to_numpy(),y.to_numpy(),model)
```

```python
model.coef_
```

```python
model.intercept_
```
