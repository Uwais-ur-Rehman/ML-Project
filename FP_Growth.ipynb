```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/groceries.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
market=[]
for i in range (0,df.shape[0]):
  cus=[]
  for j in df.columns:
    if type(df[j][i])==str:
      cus.append(df[j][i])
  market.append(cus)
```

```python
market
```

```python
l=[]
for i in market:
  for j in i:
    l.append(j)
```

```python
l
```

```python
import collections
```

```python
p=collections.Counter(l)
```

```python
d={'Item Name':p.keys(),'values':p.values()}
```

```python
pd.DataFrame(d).sort_values(by='values',ascending=False)
```

```python
from mlxtend.preprocessing.transactionencoder import TransactionEncoder
```

```python
tr=TransactionEncoder()
tr.fit(market)
```

```python
df=pd.DataFrame(tr.transform(market),columns=tr.columns_)
```

```python
df
```

```python
from mlxtend.frequent_patterns import fpgrowth
```

```python
fpgrowth(df,min_support=0.05,use_colnames=True,max_len=3).sort_values(by='support',ascending=False)
```
